##### 실습2

```python
f = open("score.txt", "w")
stu = 5
for i in range(stu):
    num = input("성적을 입력하세요")
    f.write(f"{num}\n")
f.close()
```

```python
f = open("score.txt", "r")
data=[]
for i in f:
    data.append(i)
f.close()
```

```python
f = open("report.txt", "w")
for i in data:
    line = i.split(" ")
    avg = (int(line[1])+int(line[2]))/2
    if avg >= 90:
        gra = "A"
    elif avg >= 80:
        gra = "B"
    elif avg >= 70:
        gra = "C"
    elif avg >= 60:
        gra = "D"
    else:
        gra = "F"
    f.write(f"{line[0]} {avg} ({gra})\n")
f.close()
```
