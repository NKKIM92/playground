### 실습1

```python
class Gildong:
    
    na = "A반" # 권장되는 방법은 아님
    def __init__(self, name = "길동", age = 26, height = 181, weight = 78):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight
    
    def ageplus(self):
        self.age = self.age + 1
        return self.age
    
    def heightchange(self):
        cha = int(input("키"))
        self.height = self.height + cha
        return self.height
    
    def weightchange(self):
        cha = int(input("몸무게"))
        self.weight = self.weight + cha
        return self.weight
    
    def get_info(self):
        print("반=", self.na)
        print("이름=", self.name)
        print("나이=", self.age)
        print("키=", self.height)
        print("몸무게=", self.weight)
```

```python
a = Gildong()
a.get_info()
```

```python
b = Gildong("태범", 25, 178, 75)
b.ageplus()
b.heightchange()
b.weightchange()
b.get_info()
```

### 실습 2-1

```python
class BankAccount:
    balance = 0
    name = ""
    def __init__(self, balance, name):
        self.balance = balance
        self.name = name
    
    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("잔액이 부족합니다")
    
    def get_info(self):
        print("이름:", self.name)
        print("잔고:", self.balance)
```

```python
a = BankAccount(100, "Gildong Hong")
while True:
    in1 = input("원하시는 작업을 입력하시오")
    if in1 == "입금":
        in2 = int(input("원하시는 금액을 입력하시오"))
        a.deposit(in2)
    elif in1 == "출금":
        in2 = int(input("원하시는 금액을 입력하시오"))
        a.withdraw(in2)
    elif in1 == "조회":
        a.get_info()
    elif in1 == "종료":
        print("장비를 정지합니다.")
        break
    else:
        print("잘못 입력하였습니다.")
        continue
```

```python
a = BankAccount(0, "Gildong Hong")
a.deposit(1000)
a.withdraw(900)
a.get_info()
```

### 실습 2-2

```python
class BankAccount:
    balance = 0
    name = ""
    def __init__(self, balance, name):
        self.balance = balance
        self.name = name
    
    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("잔액이 부족합니다")
    
    def transfer(self, other, amount):
        if self.withdraw(amount) == True:
            other.deposit(amount)
    
    def get_info(self):
        print("이름:", self.name)
        print("잔고:", self.balance)
```

```python
acc1 = BankAccount(1000, "고길동")
acc2 = BankAccount(0, "고희동")
acc1.transfer(acc2, 200)
acc1.get_info()
acc2.get_info()
```

### 실습3

```python
class BankAccount:
    balance = 0
    name = ""
    def __init__(self, balance, name):
        self.balance = balance
        self.name = name
    
    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print("잔액이 부족합니다")
    
    def get_info(self):
        print("이름:", self.name)
        print("잔고:", self.balance)
        
class MinimunBalanceAccount(BankAccount):
    def __init__(self, balance, name, min_bal = 0):
        super().__init__(balance, name)
        self.min_bal = min_bal
    
    def withdraw(self, amount):
        if self.balance >= self.min_bal:
            self.balance -= amount
        else:
            print("최소 잔액을 유지해야 합니다.")
```

```python
acc = MinimunBalanceAccount(500, "Kim", 1000)
acc.deposit(400)
acc.withdraw(600)
acc.get_info()
```

```python
slist = [[20180001, "Hong Gildong", 84, 73], [20180002, "Lee Jieun", 92, 89], [20180009, "Kim Cheolsu", 57, 62], [20180004, "Lee Yeonghee", 81, 84], [20180011, "Ha Donghoon", 58, 68]]
```

```python
def show(x):
    title = ["Student ","Name    ","Mid","Final","Average","Grade"]
    for i in title:
        print(i, end = "\t")
    print("-------------------------------------------------------------")
    for i in x:
        for j in i:
            print(j,end = "\t")
        print("")
        
def search(x):
    IDQ = int(input("Student ID:"))
    find = False
    for i in x:
        if IDQ == i[0]:
            title = ["Student ","Name    ","Mid","Final","Average","Grade"]
            for j in title:
                print(j, end = "\t")
            print("-------------------------------------------------------------")
            print(i)
            find = True
    if find == False:
        print("No Such Person")

for i in slist:
    avg = (int(i[2])+int(i[3]))/2
    if avg >= 90:
        gra = "A"
    elif avg >= 80:
        gra = "B"
    elif avg >= 70:
        gra = "C"
    elif avg >= 60:
        gra = "D"
    else:
        gra = "F"
    i.append(avg)
    i.append(gra)
slist.sort(key = lambda x : x[4],reverse=True)
show(slist)
search(slist)

```
