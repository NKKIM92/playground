## 실습 7번

```python
#Step1
df = pd.read_csv("FITNESS.csv", encoding = 'cp949')
```

```python
#Step2
result = df.groupby(by = ['GENDER','AGEGROUP']).mean(numeric_only=True)['WEIGHT'].reset_index()
display(result)
```

```python
#Step3
pivot_result1 = result.pivot(index = 'GENDER', columns = 'AGEGROUP', values = 'WEIGHT')
pivot_result2 = result.pivot(index = 'AGEGROUP', columns = 'GENDER', values = 'WEIGHT')
display(pivot_result1)
display(pivot_result2)
```
