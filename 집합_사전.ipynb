##### 실습1

```python
def find_participants(p1,p2):
    S1 = p1&p2
    S2 = (p1-p2)|(p2-p1)
    return S1, S2
    pass
```

```python
A = {'Alice', 'Bob', 'Charlie', 'David'}
B = {'Alice', 'Ella', 'Frank', 'Bob'}
C = {'George', 'Ella'}
both1, one1 = find_participants(A, B)
both2, one2 = find_participants(A, C)
print(both1, one1) # {'Alice', 'Bob'} {'Ella', 'David', 'Frank', 'Charlie'}
print(both2, one2) # set() {'David', 'Bob', 'Charlie', 'George', 'Alice', 'Ella'}
```

##### 실습3

```python
sentence = input("Enter a sentence:")
d = {}
for i in sentence:
    if i in d:
        d[i] = d[i] +1
    else:
        d[i] = 1
print(d)
```

```python
sentence = input("Enter a sentence:")
d = {}
for i in sentence:
    d[i] = 0
for i in sentence:
    d[i] = d[i] +1
print(d)
```

```python
sentence = input("Enter a sentence:")
b = {}
for i in sentence:
    b[i] = b.get(i,0) + 1
print(b)
```

##### 4. 변수의 unpacking

##### 실습4

```python
L1 = ['one', 'two', 'three', 'four']
L2 = [1, 2, 3, 4]
d = {}
```

```python
D = dict(zip(L1,L2))
print(D)

L = list(zip(L1,L2))
d = {}
for i in L:
    d[i[0]] = i[1]
print(d)

ANO = list(zip(L1,L2))
d = {}
for a,b in ANO:
    d[a] = b
print(d)
```
