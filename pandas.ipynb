### merge, join

```python
# 53 Page (데이터결합 연습)
# left, right 데이터 생성
left = pd.DataFrame({'key1': ['K0', 'K1', 'K2', 'K3'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
right = pd.DataFrame({'key1': ['K0', 'K1', 'K2', 'K3'],
                      'key2': ['K0', 'K0', 'K0', 'K0'],
                      'C': ['C0', 'C1', 'C2', 'C3'],
                      'D': ['D0', 'D1', 'D2', 'D3']})
```

```python
result = pd.merge(left, right, on = ['key1','key2'])
result
```

```python
result = pd.merge(left, right, on = ['key1','key2'], how = 'outer')
result
```

```python
result = pd.merge(left, right, on = ['key1','key2'], how = 'outer', indicator = True)
result
```

```python
# 57 Page (데이터결합 연습)
# left, right 데이터 생성
left = pd.DataFrame({'key1': [1, 2, 3, 4],
                     'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']})
right = pd.DataFrame({'C': ['C1', 'C2', 'C5'],
                      'D': ['D1', 'D2', 'D5']},
                     index=[1, 2, 5])
```

```python
result = left.join(right, how = 'inner')
result
```

```python
result = left.join(right, how = 'right')
result
```

```python
import numpy as np
import pandas as pd
pop_seoul = pd.read_csv("서울인구통계.txt", sep = "\t", thousands = ',', skiprows = 2, encoding = 'utf-8')
pop_seoul.head()
```

```python
pop1 = pop_seoul[['자치구', '계', '계.1', '계.2', '65세이상고령자']]
pop_seoul=pop1.drop([0])
pop_seoul.head()
```

```python
pop_seoul.rename(columns = {pop_seoul.columns[0]:'구별',
                            pop_seoul.columns[1]:'인구수',
                            pop_seoul.columns[2]:'한국인',
                            pop_seoul.columns[3]:'외국인',
                            pop_seoul.columns[4]:'고령자'}, inplace = True)
pop_seoul.head()
```

```python
cctv_seoul = pd.read_excel("서울시CCTV.xlsx", skiprows= 2, names= ['기관명', '소계', '2011년 이전', \
 '2012년', '2013년', '2014년', '2015년', '2016년', '2017년', '2018년', '2019년', '2020년', '2021년'])
cctv_seoul.head()
            
```

```python
cctv_seoul.rename(columns = {cctv_seoul.columns[0]: '구별'}, inplace = True)
cctv_seoul.rename(columns = {cctv_seoul.columns[1]: 'cctv수'}, inplace = True)
cctv_seoul.head()
```

```python
cctv_seoul.sort_values(by = 'cctv수', ascending= True)
```

```python
cctv_seoul['최근증가율']=(cctv_seoul['2021년']+cctv_seoul['2020년']+cctv_seoul['2019년']) \
   /(cctv_seoul['cctv수']-(cctv_seoul['2021년']+cctv_seoul['2020년']+cctv_seoul['2019년']))*100
cctv_seoul.sort_values(by='최근증가율',ascending=False).head(5)
```

```python
pop_seoul['외국인 비율'] = pop_seoul['외국인']/pop_seoul['인구수']*100
pop_seoul['고령자 비율'] = pop_seoul['고령자']/pop_seoul['인구수']*100
pop_seoul.head()
```

```python
pop_seoul.sort_values(by = '외국인 비율')
```

```python
pop_seoul.sort_values(by = '고령자 비율')
```

```python
cctv_seoul['구별'] = cctv_seoul['구별'].str.replace(' ','')
result = pd.merge(cctv_seoul, pop_seoul, on= '구별')
result
```
