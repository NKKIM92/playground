### 실습1

```python
import numpy as np
arr = np.zeros((5,5))
for i in range(5):
    if i % 5 == 0 or i % 5 == 4:
        arr[i] = 1
    for j in range(5):
        if j % 5 == 0 or j % 5 == 4:
            arr[i][j] = 1
print(arr)
```

```python
import numpy as np
arr = np.ones((5,5))
arr[1:-1,1:-1] = 0
print(arr)
```

```python
import numpy as np
arr = np.ones((3,3))
arr = np.pad(arr, pad_width = 1, mode = 'constant', constant_values = 0)
print(arr)
```

### 실습2

```python
import numpy as np
arr = np.zeros((8,8))
for i in range(8):
    if i % 2 == 1:
        for j in range(8):
            if j % 2 ==0:
                arr[i][j] = 1
    else:
        for j in range(8):
            if j % 2 ==1:
                arr[i][j] = 1
print(arr)
```

```python
import numpy as np
arr = np.ones((8,8))
arr[1::2,1::2] = 0
arr[::2,::2] = 0
print(arr)
```
